```python
from sklearn.datasets import make_moons

moons = make_moons(n_samples=10000, noise=0.4)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(moons[0], moons[1], test_size=0.2, random_state=42)
```

```python
from sklearn.model_selection import ShuffleSplit

mini_sets = []

rs = ShuffleSplit(n_splits=1000, train_size=100, random_state=42)

for mini_train_index, mini_test_index in rs.split(X_train):
    X_mini_train = X_train[mini_train_index]
    y_mini_train = y_train[mini_train_index]
    mini_sets.append((X_mini_train, y_mini_train))
```

```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

preds_grid = np.zeros((len(y_test), 1000))

for i in range(1000):
    X_curr, y_curr = mini_sets[i]

    model = DecisionTreeClassifier(max_depth=10, max_leaf_nodes=20, random_state=42)

    model.fit(X_curr, y_curr)

    preds = model.predict(X_test)
    preds_grid[:, i] = preds

    if(i in [0,1,2,3,4]):
        print(f"accuracy of tree {i}: {accuracy_score(y_test, preds)}")
```

```python
from scipy.stats import mode

final_preds = mode(preds_grid, axis=1).mode.ravel()

accuracy_score(y_test, final_preds)
```
